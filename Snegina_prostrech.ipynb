```python
import re
import csv

COLLOQUIAL_LEXICON = {

    "жисть", "по́читай", "приятственны", "тягот", "напасти", "стягловица",
    "пришлеца", "заневестилася", "охлопья", "тигулёвка", "сноровка",
    "водь", "выть", "гнусь", "ноять", "дури", "хлюпь", "чё", "чо",
    "трюсь", "трьясь", "трясь", "пересуд", "курево", "дуро́вая", "зыкь",
    "приволок", "гумно", "чадо", "чуя", "чую", "по́чуял", "даден",
    "бе́чь", "прясло", "голытьба", "запетлю", "рюшка", "девки",
    "зыбистый", "жись", "встрел", "ливисто", "ливенка", "разухабистой",
    "зыками", "остальние", "ходя", "напузырить", "брехун", "пoнюшка",
    "скочили", "зуботычины", "поёмы", "сычёная", "забоченились",
    "запенились", "гать", "в прутнику", "кивливом", "сучья", "ухлюпы",
    "ухваты", "дупеля", "старуха", "Прон", "Сергуха", "Снегина",
    "галдеж", "жисть", "оглоушу", "бряк", "прикокошить", "на помо́г",
    "помо́г", "слыхать", "услы́хал", "каза́ть", "чадо", "небось",
    "корогод", "летошней", "ворожбинные", "купырях", "хошь", "чикнули",
    "не кажешься", "прижваривать", "дуро́вая", "расея", "зыкь",
    "помело", "жилец", "мелево", "лещуга", "на закорки", "путю",
    "посолонка", "забольная", "подожочек", "подтыками", "кудашний",
    "бурыга", "чапыга", "суровика", "пьяника", "стягловица", "ухват"
}

COLLOQUIAL_LEXICON = {w.lower().replace('́', '') for w in COLLOQUIAL_LEXICON}

def load_text(path):
    with open(path, 'r', encoding='utf-8') as f:
        return f.read()

text = load_text("/content/Esenin_Anna_Snegina_.txt")

def split_sentences(text):

    return re.split(r'(?<=[.!?])\s+(?=[А-Я"«–])', text)

sentences = split_sentences(text)

results = []

for sent in sentences:
    sent_clean = sent.strip()
    if not sent_clean:
        continue


    search_sent = sent_clean.lower().replace('́', '')

    found_words = []
    for word in COLLOQUIAL_LEXICON:

        if re.search(r'\b' + re.escape(word) + r'\b', search_sent):
            found_words.append(word)

    if found_words:
        is_dialogue = bool(re.search(r'[«"]|–\s*[А-Я]', sent_clean))
        results.append({
            "sentence": sent_clean,
            "colloquial_words": "; ".join(found_words),
            "is_dialogue": "да" if is_dialogue else "нет"
        })

output_path = "/content/anna_snegina_colloquial_detected.csv"
with open(output_path, "w", encoding="utf-8-sig", newline='') as f:
    writer = csv.DictWriter(f, fieldnames=["sentence", "colloquial_words", "is_dialogue"])
    writer.writeheader()
    writer.writerows(results)

print(f"Нашел {len(results)} предложений с просторечной лексикой.")
print(f"Результат сохранён: {output_path}")

print("\nПримеры:")
for r in results[:5]:
    print(f"- {r['sentence'][:80]}... → [{r['colloquial_words']}]")
```
